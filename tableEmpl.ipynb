```python
import numpy as np
import pandas as pd
```

```python
# Employee data
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Aman", "Priya", "Vikas", "Neha"],
    "Age": [25, 28, 24, 22, 26],
    "Gender": ["Male", "Male", "Female", "Male", "Female"],
    "Department": ["IT", "HR", "Finance", "IT", "Marketing"],
    "Salary": [35000, 45000, 30000, 25000, 40000],
    "City": ["Delhi", "Mumbai", "Lucknow", "Kanpur", "Noida"],
    "Experience": [2, 5, 1, 0, 3]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display table
print(df)
```

```python
# Display first 3 records
print(df.head(3))

# Display last 2 records
print(df.tail(2))

# Display column names
print(df.columns)

# Display information
print(df.info())

# Display statistical summary
print(df.describe())

# Find employees with salary above 30000
print(df[df["Salary"] > 30000])

# Sort employees by salary
print(df.sort_values("Salary"))

# Add a new column
df["Bonus"] = df["Salary"] * 0.10

# Calculate average salary
print(df["Salary"].mean())

# Save data into CSV
df.to_csv("employees.csv", index=False)
```

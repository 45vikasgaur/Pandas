```python
import pandas as pd

data = {
    'Emp_ID': [201, 202, 203, 204, 205, 206],
    'Name': ['Aman', 'Ravi', 'Priya', 'Sneha', 'Karan', 'Anjali'],
    'Department': ['IT', 'HR', 'Finance', 'IT', 'Sales', 'HR'],
    'Age': [23, 28, 25, 30, 27, 24],
    'Salary': [30000, 45000, 35000, 60000, 40000, 32000],
    'Experience': [1, 4, 2, 6, 3, 1],
    'City': ['Lucknow', 'Delhi', 'Mumbai', 'Noida', 'Pune', 'Kanpur']
}

df = pd.DataFrame(data)

print(df)
```

```python
# Addition
print(df['Salary'] + 5000)

# Subtraction
print(df['Salary'] - 5000)

# Multiplication
print(df['Salary'] * 2)

# Division
print(df['Salary'] / 2)

# Modulus
print(df['Salary'] % 10000)

# Exponentiation
print(df['Age'] ** 2)
```

```python
# Equal to
print(df['Department'] == 'IT')

# Not equal to
print(df['Department'] != 'HR')

# Greater than
print(df['Salary'] > 35000)

# Less than
print(df['Age'] < 25)

# Greater than or equal to
print(df['Experience'] >= 3)

# Less than or equal to
print(df['Salary'] <= 40000)
```

```python
# AND operator
print(df[(df['Age'] > 24) & (df['Salary'] > 35000)])

# OR operator
print(df[(df['Department'] == 'IT') | (df['Department'] == 'HR')])

# NOT operator
print(df[~(df['Department'] == 'IT')])
```

```python
df['Salary'] += 5000
print(df)

df['Age'] += 1
print(df)

df['Salary'] -= 2000
print(df)
```

```python
# Check whether IT exists in Department
print('IT' in df['Department'].values)

# Check whether Sales exists
print('Sales' not in df['Department'].values)
```

```python
# Find maximum salary
print(df['Salary'].max())

# Find minimum salary
print(df['Salary'].min())

# Find total salary
print(df['Salary'].sum())

# Find average salary
print(df['Salary'].mean())

# Sort employees by salary
print(df.sort_values(by='Salary', ascending=False))

# Group employees by department
print(df.groupby('Department')['Salary'].mean())
```

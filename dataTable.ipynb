```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Rahul', 'Amit', 'Priya', 'Neha', 'Rohit'],
    'Department': ['IT', 'HR', 'Finance', 'IT', 'Marketing'],
    'Age': [25, 28, 24, 30, 27],
    'Salary': [35000, 40000, 32000, 50000, 38000],
    'City': ['Delhi', 'Lucknow', 'Mumbai', 'Noida', 'Pune']
}

df = pd.DataFrame(data)

print(df)
```

```python
# Display first 5 rows
print(df.head())

# Display last 5 rows
print(df.tail())

# Display dataset information
print(df.info())

# Statistical summary
print(df.describe())

# Display column names
print(df.columns)

# Find employees in IT department
print(df[df['Department'] == 'IT'])

# Find employees with salary above 35000
print(df[df['Salary'] > 35000])

# Calculate average salary
print(df['Salary'].mean())

# Department-wise employee count
print(df['Department'].value_counts())
```

```python
df.to_csv('organisation.csv', index=False)
```
